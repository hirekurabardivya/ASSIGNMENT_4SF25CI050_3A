5. Data Aggregation 
Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category

```python
df=pd.DataFrame({
    'category':['a','b','c'],
    'value':[1,2,3]
})

df
```

```python
df1=df.groupby('category')['value'].agg(['sum','mean'])
print(df1)
```

8. MultiIndex DataFrame
Create a Pandas DataFrame with a MultiIndex (hierarchical index). Perform some basic indexing and slicing operations on the MultiIndex DataFrame.

```python
import pandas as pd
index = pd.MultiIndex.from_tuples([
    (2025, 'Q1'),
    (2025, 'Q2'),
    (2026, 'Q1'),
    (2026, 'Q2')
], names=['Year', 'Quarter'])
df = pd.DataFrame({'Sales': [100, 150, 200, 250]}, index=index)
print("--- DataFrame ---")
print(df)
print("\n")
print("--- Select Year 2025 ---")
print(df.loc[2025])
print("\n")
print("--- Select 2025 Q2 ---")
print(df.loc[(2025, 'Q2')])
print("\n")
df = df.sort_index()
print("--- Slice from 2025 Q2 to 2026 Q1 ---")
print(df.loc[(2025, 'Q2'):(2026, 'Q1')])
```

9. Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'. Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.

```python
import pandas as pd
import numpy as np
index = pd.MultiIndex.from_tuples(
    [('Fruit', 'Apple'), ('Fruit', 'Mango'),
     ('Vegetable', 'Carrot'), ('Vegetable', 'Potato')],
    names=['Category', 'SubCategory']
)
df = pd.DataFrame({'Value': np.random.randint(1, 100, 4)}, index=index)
print("DataFrame:")
print(df)
print("\nSum by Category:")
print(df.groupby(level='Category').sum())
print("\nSum by SubCategory:")
print(df.groupby(level='SubCategory').sum())
```

11. Create a Pandas DataFrame with 3 columns and 6 rows filled with random integers. Apply a lambda function to create a new column that is the sum of the existing columns.


```python
import numpy as np
df=pd.DataFrame(
    np.random.randint(1,10,(6,3)),
    columns=['a','b','c']
)
df
```
