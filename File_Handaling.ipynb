1. Counting Words in a File
Write a function that reads the contents of a file named `document.txt` and returns the number of words in the file.

```python
with open("document.txt","w") as file:
    file.write("Hello World!\n")
    file.write("good morning!\n")  
with open("document.txt","r") as file:
    content=file.read()
    print(content)
```

```python
words=content.split()
print("number of words: ",len(words))
```

2. Finding and Replacing Text
Write a function that finds and replaces all occurrences of a given word in a file named `data.txt` with another word.

```python
with open("data.txt","w")as file:
    file.write("Hello World!\n")
    file.write("Hello Everyone\n")
    file.write("Good Morning!\n")
with open("data.txt","r")as file :
    content=file.read()
    print(content)

    
```

```python
def find_replace(old,new):
    with open("data.txt","r")as file :
        content=file.read()
    new_content=content.replace(old,new)
    with open("new_data.txt","w")as file:
        file.write(new_content)
        print(new_content)
old="Hello"
new="Hi"
find_replace(old,new)
```

3. Reading a File in Reverse
Write a function that reads the contents of a file named `reverse.txt` and prints each line in reverse order.

```python
with open("reverse.txt","w")as file:
    file.write("Hello World!\n")
    file.write("Hello Everyone\n")
    file.write("Good Morning!\n")
with open("reverse.txt","r")as file :
    content=file.read()
    print(content) 
```

```python
with open("reverse.txt","r")as file:
    lines=file.readlines()
    for line in reversed(lines):
        print(line,end="")
```
